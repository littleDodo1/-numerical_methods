```python
import math
import matplotlib.pyplot as plt
import numpy as np


class NumericalDiffTaskVariant13:

    def __init__(self):
        self.X_star = 0.8
        
        self.nodes = [0.2, 0.5, 0.8, 1.1, 1.4]
        self.y_vals = [12.9060, 5.5273, 3.8777, 3.2692, 3.0319]


def compute_finite_differences(y_vals):

    n = len(y_vals)
    diff_table = [[0.0] * n for _ in range(n)]
    
    for i in range(n):
        diff_table[i][0] = y_vals[i]
        
    for j in range(1, n):
        for i in range(n - j):
            diff_table[i][j] = diff_table[i + 1][j - 1] - diff_table[i][j - 1]
            
    return diff_table


def calculate_derivatives(nodes, y_vals, target_x):

    h = nodes[1] - nodes[0]
    idx = nodes.index(target_x)
    diff_table = compute_finite_differences(y_vals)

    d1_left = diff_table[idx - 1][1] / h if idx > 0 else None
    d1_right = diff_table[idx][1] / h if idx < len(nodes) - 1 else None
    d1_central = (diff_table[idx - 1][1] + diff_table[idx][1]) / (2 * h) if (0 < idx < len(nodes) - 1) else None
    d2 = diff_table[idx - 1][2] / (h ** 2) if (0 < idx < len(nodes) - 1) else None
    
    used_nodes = [nodes[idx - 1], nodes[idx], nodes[idx + 1]] if (0 < idx < len(nodes) - 1) else nodes
    
    return {
        'h': h,
        'idx': idx,
        'd1_left': d1_left,
        'd1_right': d1_right,
        'd1_central': d1_central,
        'd2': d2,
        'used_nodes': used_nodes
    }


def main():
    task = NumericalDiffTaskVariant13()
    res = calculate_derivatives(task.nodes, task.y_vals, task.X_star)
    
    print("=" * 95)
    print(" ЧИСЛЕННОЕ ДИФФЕРЕНЦИРОВАНИЕ")
    print("=" * 95)
    
    print("\n" + "-" * 95)
    print("1. ИСХОДНАЯ ТАБЛИЦА ЗНАЧЕНИЙ ФУНКЦИИ")
    print("-" * 95)
    print(f"{'i':<4} | {'x_i':<16} | {'y_i = f(x_i)':<16}")
    print("-" * 95)
    for i in range(len(task.nodes)):
        print(f"{i:<4} | {task.nodes[i]:<16.4f} | {task.y_vals[i]:<16.4f}")
        
    # 2. Точка, в которой выполняется вычисление
    print("\n" + "-" * 95)
    print("2. ТОЧКА ВЫЧИСЛЕНИЯ ПРОИЗВОДНЫХ")
    print("-" * 95)
    print(f"  Заданная контрольная точка X* = {task.X_star:.4f} (узел i = {res['idx']})")
    print(f"  Шаг сетки h = {res['h']:.4f}")
    
    # 3. Узлы таблицы, использованные при вычислении производных
    print("\n" + "-" * 95)
    print("3. УЗЛЫ ТАБЛИЦЫ, ИСПОЛЬЗОВАННЫЕ ПРИ ВЫЧИСЛЕНИИ ПРОИЗВОДНЫХ")
    print("-" * 95)
    print(f"  Использованные узлы x_i: {res['used_nodes']}")
    print(f"  Соответствующие y_i:     {[task.y_vals[task.nodes.index(x)] for x in res['used_nodes']]}")
    
    # 4 & 5. Значения первой и второй производных
    print("\n" + "=" * 95)
    print("4 & 5. РЕЗУЛЬТАТЫ ВЫЧИСЛЕНИЯ ПРОИЗВОДНЫХ В ТОЧКЕ X* = 0.8")
    print("=" * 95)
    print(f"  1. Левосторонняя первая производная f'_лево(X*):  {res['d1_left']:12.6f}")
    print(f"  2. Правосторонняя первая производная f'_право(X*): {res['d1_right']:12.6f}")
    print(f"  3. Центральная первая производная f'_центр(X*):   {res['d1_central']:12.6f}  (порядок O(h^2))")
    print(f"  4. Вторая производная f''(X*):                     {res['d2']:12.6f}  (порядок O(h^2))")
    print("=" * 95)


if __name__ == '__main__':
    main()
```
